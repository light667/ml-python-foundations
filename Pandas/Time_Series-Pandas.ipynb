```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
bitcoin = pd.read_csv('BTC-EUR.csv', index_col='Date', parse_dates= True)
```

```python
bitcoin.head()
```

```python
bitcoin['Close'].plot(figsize=(9, 6))
plt.show()
```

```python
#bitcoin['2010']['Close'].plot()
```

```python
bitcoin.loc['2019', 'Close'].resample('M').mean().plot()
plt.show()
```

```python
bitcoin.loc['2019', 'Close'].resample('W').agg(['mean', 'std', 'min']).plot()
plt.show()
```
